```python
df2=pd.read_csv('https://raw.githubusercontent.com/krishnaik06/playstore-Dataset/main/googleplaystore.csv')
```

```python
df_copy = df_copy.loc[:, ~df_copy.columns.duplicated()]
```

```python
df_copy = pd.concat([df_copy,df2['Price']],axis =1)
```

```python
df_copy
```

```python
duplicate_columns = df_copy.columns[df_copy.columns.duplicated()].unique()
```

```python
duplicate_columns
```

```python
for col in duplicate_columns:
    matching_cols = df_copy.loc[:, df_copy.columns == col]
    if matching_cols.shape[1] > 1:
        # Loop through duplicates to find and drop the all-NaN version
        for i in range(matching_cols.shape[1]):
            if matching_cols.iloc[:, i].isna().all():
                df_copy.drop(columns=matching_cols.columns[i], inplace=True)
                break  # Drop only one if there's exactly one with NaN
```

```python
df_copy['Last Updated'] = pd.to_datetime(df_copy['Last Updated'])
```

```python
# Convert 'last_updated' to datetime, mark errors as NaT
df_copy['Last Updated'] = pd.to_datetime(df_copy['Last Updated'], errors='coerce')

# Drop rows where conversion failed (i.e., rows with invalid dates)
df_copy = df_copy.dropna(subset=['Last Updated'])
```

```python
df_copy['Day'] = df_copy['Last Updated'].dt.day
df_copy['Month'] = df_copy['Last Updated'].dt.month
df_copy['Year'] = df_copy['Last Updated'].dt.year
```
