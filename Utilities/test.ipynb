```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Possible outcomes
X = np.array([1, 2, 3, 4, 5, 6])
P_X = np.ones(6) / 6  # Each outcome has equal probability

# Plot the PMF
plt.figure(figsize=(8,5))
sns.barplot(x=X, y=P_X, color='blue', alpha=0.7)
plt.xlabel("Outcome (X)")
plt.ylabel("P(X)")
plt.title("PMF of a Fair 6-Sided Die")
plt.show()
```

```python
from scipy.stats import binom
import matplotlib.pyplot as plt

# Parameters
n = 10      # number of trials
p = 0.5     # probability of success

# PMF (Probability Mass Function)
x = range(0, n+1)
pmf = binom.pmf(x, n, p)

# Plot the distribution
plt.bar(x, pmf)
plt.title('Binomial Distribution (n=10, p=0.5)')
plt.xlabel('Number of Successes')
plt.ylabel('Probability')
plt.show()
```

```python
import numpy as np
import pandas as pd
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns

# Example 1: Chi-Squared Test of Independence
# Creating a contingency table
data = pd.DataFrame({
    'Gender': ['Male', 'Male', 'Female', 'Female', 'Male', 'Female', 'Male', 'Female', 'Female', 'Male'],
    'Product_Preference': ['Product_A', 'Product_B', 'Product_A', 'Product_A', 'Product_B', 'Product_B', 'Product_A', 'Product_B', 'Product_A', 'Product_B']
})

# Create the contingency table
contingency_table = pd.crosstab(data['Gender'], data['Product_Preference'])
print("Contingency Table:")
print(contingency_table)

# Perform chi-squared test
chi2, p, dof, expected = stats.chi2_contingency(contingency_table)

print(f"\nChi-squared statistic: {chi2:.4f}")
print(f"p-value: {p:.4f}")
print(f"Degrees of freedom: {dof}")
print("Expected frequencies:")
print(pd.DataFrame(expected, index=contingency_table.index, columns=contingency_table.columns))
```
