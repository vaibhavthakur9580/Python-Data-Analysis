```python
import pandas as pd
import seaborn as sns
tips = sns.load_dataset('tips')
```

```python
#label encoding 
sex_map = {'Male': 0, 'Female':1}
tips['sex_encoding'] = tips['sex'].map(sex_map)
print(tips[['sex', 'sex_encoding']].head())
```

```python
#one hot encoding 
from sklearn.preprocessing import OneHotEncoder
encoder = OneHotEncoder()
tips_df = pd.DataFrame(tips['sex'])
tips_encoded = encoder.fit_transform(tips_df['sex']).toarray()
tips_encoded
```

```python
tips_onehot_encoded = pd.get_dummies(tips, columns=['sex'],prefix='sex', drop_first=True)
```
